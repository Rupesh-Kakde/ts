```python
from flask import Flask, abort

app = Flask(__name__)

@app.route('/')
def home():
    return 'Welcome to Flask'

@app.route('/bad')
def bad_request():
    abort(400)

@app.route('/forbidden')
def forbidden():
    abort(403)

if __name__ == '__main__':
    app.run()
```

```python
from flask import Flask

app = Flask(__name__)

@app.route('/table/<int:num>')
def table(num):
    result = ""

    for i in range(1, 11):
        result += f"{num} x {i} = {num * i}<br>"

    return result

app.run(port=5002)
```
